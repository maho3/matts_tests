```python
%load_ext autoreload
%autoreload 2
%matplotlib inline

import os
from os.path import join
import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl
import Pk_library as PKL
from copy import deepcopy
mpl.style.use('./style.mcstyle')
# os.chdir('..')
```

```python
from cmass.bias.apply_hod import load_snapshot, populate_hod, parse_hod
from cmass.utils import load_params
from omegaconf import OmegaConf
```

```python
lhid = 0
simpath = f'/anvil/scratch/x-mho1/cmass-ili/quijotelike/fastpm/L1000-N128/{lhid}'
hpos, hvel, hmass, hmeta = load_snapshot(simpath, 0.666667)
cosmo = load_params(lhid, '../params/latin_hypercube_params.txt')
conf = OmegaConf.load(f'{simpath}/config.yaml')
model = 'leauthaud11'
conf.bias.hod.model = model
conf.bias.hod.seed = 1
conf.bias.hod.default_params = 'behroozi10'
conf = parse_hod(conf)
print(OmegaConf.to_yaml(conf))
```

### Generic test

```python
L = 1000
N = 128
zf = 0.5
# model = 'zheng07'
theta = conf.bias.hod.theta
seed = 0
mdef = 'vir'
hod = populate_hod(
    hpos, hvel, hmass,
    cosmo, L, zf,
    model, theta,
    hmeta,
    seed, 'vir'
)

gpos = np.array([hod['x'], hod['y'], hod['z']]).T
gvel = np.array([hod['vx'], hod['vy'], hod['vz']]).T
print(gpos.shape)
```

```python
hhist, _ = np.histogramdd(
    hpos, bins=[128]*3,
    range=[(0, L)]*3)

ghist, edges = np.histogramdd(
    gpos, bins=[128]*3,
    range=[(0, L)]*3)

f, axs = plt.subplots(1, 2, figsize=(12, 6))
axs[0].imshow(hhist[:10].mean(axis=0), origin='lower')
axs[0].set_title('Halos')
axs[1].imshow(ghist[:10].mean(axis=0), origin='lower')
axs[1].set_title('Galaxies')

for ax in axs:
    ax.axis('off')
```

```python
hx = (hhist/np.mean(hhist) - 1).astype(np.float32)
gx = (ghist/np.mean(ghist) - 1).astype(np.float32)
XPk = PKL.XPk([hx, gx], BoxSize=L, axis=0, MAS=['CIC', 'CIC'], threads=1)
XPk2 = PKL.XPk([hx, gx], BoxSize=L, axis=0, MAS=['CIC', 'CIC'], threads=1)

knyq = np.pi*N/L

f, axs = plt.subplots(1, 2, figsize=(12, 6))
ax = axs[0]
ax.plot(XPk.k3D, XPk.Pk[:, 0, 0], label='halos')
ax.plot(XPk.k3D, XPk.Pk[:, 0, 1], label='galaxies (c-M)')
ax.plot(XPk2.k3D, XPk2.Pk[:, 0, 1], label='galaxies (CHARM)')
ax.axvline(knyq, color='r', ls='--')
ax.set(xlabel='k', ylabel='P(k)')
ax.legend()
ax.loglog()
ax.set_title('Power Spectrum')

ax = axs[1]
ax.axhline(1, color='k', ls='--')
ax.plot(XPk.k3D, XPk.XPk[:, 0, 0] /
        np.sqrt((XPk.Pk[:, 0, 0]*XPk.Pk[:, 0, 1])), c='C1')
ax.plot(XPk2.k3D, XPk2.XPk[:, 0, 0] /
        np.sqrt((XPk2.Pk[:, 0, 0]*XPk2.Pk[:, 0, 1])), c='C2')
ax.axvline(knyq, color='r', ls='--')
ax.semilogx()
ax.set(xlabel='k', ylabel='cross-P(k)')
ax.set_title('Cross Power Spectrum')
```

### Fixed HOD Parameters

```python
Nsamp = 100
XPks = []

for i in range(Nsamp):
    hod = populate_hod(hpos, hvel, hmass, cosmo, conf, 0)

    gpos = np.array([hod['x'], hod['y'], hod['z']]).T
    ghist, edges = np.histogramdd(
        gpos, bins=[256]*3,
        range=[(0, 2000)]*3)
    gx = (ghist/np.mean(ghist) - 1).astype(np.float32)
    XPk = PKL.XPk([hx, gx], BoxSize=2000, axis=0, MAS=['CIC', 'CIC'], threads=1)
    XPks.append(XPk)
```

```python
a = [XPk.Pk[:, 0, 1]for XPk in XPks]
amu, astd = np.mean(a, axis=0), np.std(a, axis=0)
b = [XPk.XPk[:, 0, 0]/np.sqrt((XPk.Pk[:, 0, 0]*XPk.Pk[:, 0, 1]))
     for XPk in XPks]
bmu, bstd = np.mean(b, axis=0), np.std(b, axis=0)
```

```python
hx = (hhist/np.mean(hhist) - 1).astype(np.float32)
gx = (ghist/np.mean(ghist) - 1).astype(np.float32)
XPk = PKL.XPk([hx, gx], BoxSize=2000, axis=0, MAS=['CIC', 'CIC'], threads=1)

knyq = np.pi*256/2000

f, axs = plt.subplots(1, 2, figsize=(12, 6))
ax = axs[0]
ax.plot(XPk.k3D, XPk.Pk[:, 0, 0], c='k', label='halos')
ax.plot(XPk.k3D, amu, label='galaxies')
ax.fill_between(XPk.k3D, amu-astd, amu+astd, alpha=0.3)
ax.axvline(knyq, color='r', ls='--')
ax.set(xlabel='k', ylabel='P(k)')
ax.legend()
ax.loglog()
ax.set_title('Power Spectrum')

ax = axs[1]
ax.axhline(1, color='k', ls='--')
ax.plot(XPk.k3D, bmu)
ax.fill_between(XPk.k3D, bmu-bstd, bmu+bstd, alpha=0.3)
ax.axvline(knyq, color='r', ls='--')
ax.semilogx()
ax.set(xlabel='k', ylabel='cross-P(k)')
ax.set_title('Cross Power Spectrum')
```

### Scanning HOD parameters

```python
Nsamp = 10
minval, maxval = 13, 15
XPks = []

hx = (hhist/np.mean(hhist) - 1).astype(np.float32)
for i, r in enumerate(np.linspace(minval, maxval, Nsamp)):
    conf.bias.hod.theta.logMmin = float(r)

    hod = populate_hod(hpos, hvel, hmass, cosmo, conf, 0)

    gpos = np.array([hod['x'], hod['y'], hod['z']]).T
    ghist, edges = np.histogramdd(
        gpos, bins=[256]*3,
        range=[(0, 2000)]*3)
    gx = (ghist/np.mean(ghist) - 1).astype(np.float32)
    XPk = PKL.XPk([hx, gx], BoxSize=2000, axis=0, MAS=['CIC', 'CIC'], threads=1)
    XPks.append(XPk)
```

```python
hx = (hhist/np.mean(hhist) - 1).astype(np.float32)
gx = (ghist/np.mean(ghist) - 1).astype(np.float32)
XPk = PKL.XPk([hx, gx], BoxSize=2000, axis=0, MAS=['CIC', 'CIC'], threads=1)

knyq = np.pi*256/2000
cmap = plt.get_cmap('viridis')

f, axs = plt.subplots(1, 2, figsize=(12, 6))
ax = axs[0]
ax.plot(XPk.k3D, XPk.Pk[:, 0, 0], c='k', label='halos')
for i, XPk in enumerate(XPks):
    ax.plot(XPk.k3D, XPk.Pk[:, 0, 1], c=cmap(i/Nsamp))
ax.axvline(knyq, color='r', ls='--')
ax.set(xlabel='k', ylabel='P(k)')
ax.legend()
ax.loglog()
ax.set_title('Power Spectrum')

ax = axs[1]
ax.axhline(1, color='k', ls='--')
for i, XPk in enumerate(XPks):
    ax.plot(XPk.k3D, XPk.XPk[:, 0, 0]/np.sqrt((XPk.Pk[:, 0, 0]*XPk.Pk[:, 0, 1])),
            c=cmap(1-i/Nsamp), label=f'{minval + i*(maxval-minval)/(Nsamp-1):.2f}')
ax.axvline(knyq, color='r', ls='--')
ax.semilogx()
ax.legend(title='logMmin')
ax.set(xlabel='k', ylabel='cross-P(k)')
ax.set_title('Cross Power Spectrum')
```

### Test different redshift slices

```python
lhid = 0
simpath = f'/anvil/scratch/x-mho1/cmass-ili/quijotelike/fastpm/L1000-N128/{lhid}'
hpos, hvel, hmass, hmeta = load_snapshot(simpath, 0.666667)
cosmo = load_params(lhid, '../params/latin_hypercube_params.txt')
conf = OmegaConf.load(f'{simpath}/config.yaml')
print(OmegaConf.to_yaml(conf))
```

```python
from cmass.utils import cosmo_to_astropy
L = 1000
N = 128
zf = 0.3
# model = 'zheng07'
theta = conf.bias.hod.theta
cosmo = cosmo_to_astropy(cosmo)
seed = 0
mdef = 'vir'
hod = populate_hod(
    hpos, hvel, hmass,
    cosmo, L, zf,
    model, theta,
    hmeta,
    seed, 'vir'
)

gpos = np.array([hod['x'], hod['y'], hod['z']]).T
gvel = np.array([hod['vx'], hod['vy'], hod['vz']]).T
print(gpos.shape)
```

```python
from cmass.diagnostics.tools import MA, MAz, calcPk
L, N = 1000, 128
MAS = 'NGP'
delta = MA(gpos, L, N, MAS)
deltaz = MAz(gpos, gvel, L=L, N=N, MAS=MAS, z=0.5, cosmo=cosmo)

k, Pk = calcPk(delta, L, MAS=MAS)
```

```python
zflist = np.arange(0.3, 0.8, 0.1)

f, axs = plt.subplots(3, 2, figsize=(12, 12), sharex=True)
for i, zf in enumerate(zflist):
    L = 1000
    N = 128
    # zf = 0.3
    # model = 'zheng07'
    theta = conf.bias.hod.theta
    from cmass.utils import cosmo_to_astropy
    cosmo = cosmo_to_astropy(cosmo)
    seed = 0
    mdef = 'vir'
    hod = populate_hod(
        hpos, hvel, hmass,
        cosmo, L, zf,
        model, theta,
        hmeta,
        seed, 'vir'
    )

    gpos = np.array([hod['x'], hod['y'], hod['z']]).T
    gvel = np.array([hod['vx'], hod['vy'], hod['vz']]).T

    from cmass.diagnostics.tools import MA, MAz, calcPk
    L, N = 1000, 128
    MAS = 'NGP'
    delta = MA(gpos, L, N, MAS)
    deltaz = MAz(gpos, gvel, L=L, N=N, MAS=MAS, z=0.5, cosmo=cosmo)

    k, Pk = calcPk(delta, L, MAS=MAS)
    for j, ax in enumerate(axs[:, 0]):
        if j == 0:
            ax.loglog()
            ax.plot(k, Pk[:, j], label=f'zf={zf:.2f}')
            ax.set_ylabel('P0')
        else:
            ax.semilogx()
            ax.plot(k, Pk[:, j]/Pk[:, 0], label=f'zf={zf:.2f}')
            ax.set_ylabel(f'P{j}/P0')
    k, Pk = calcPk(deltaz, L, MAS=MAS)
    for j, ax in enumerate(axs[:, 1]):
        if j == 0:
            ax.loglog()
            ax.plot(k, Pk[:, j], label=f'zf={zf:.2f}')
        else:
            ax.semilogx()
            ax.plot(k, Pk[:, j]/Pk[:, 0], label=f'zf={zf:.2f}')
ax.legend()
axs[0, 0].set_title('Real Space')
axs[0, 1].set_title('Redshift Space')
axs[-1, 0].set_xlabel('k')
axs[-1, 1].set_xlabel('k')
```
